```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

from ignite.metrics.rec_sys import NDCG

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

import gzip
import ast
from pathlib import Path
```

```python
project_dir = Path("..")
comparison = pd.read_csv(project_dir / "data" / "processed" / "game_comparison.csv")

user_games = pd.read_parquet(project_dir / "data" / "processed" / "user_games.parquet")
```

```python
user_games.info()
```

```python
def recommend_popular_games(user_id, n=10):
    owned_games = set(
        user_games.loc[user_games["user_id"] == user_id, "item_id"]
    )

    recommendations = (
        comparison[~comparison["item_id"].isin(owned_games)]
        .sort_values("users_count", ascending=False)
        .drop_duplicates("item_id")
        .head(n)
    )

    return recommendations[["item_id", "item_name", "users_count"]]
```

```python
example_user = user_games["user_id"].iloc[0]
recommend_popular_games(example_user, n=10)
```
